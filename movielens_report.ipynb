# Аналитический отчёт по данным MovieLens
В данном отчёте проведено комплексное исследование уменьшенного набора данных MovieLens. 
Мы проанализируем кинематографические тренды, предпочтения пользователей, специфику текстовых тегов, а также получим финансовые и производственные метрики фильмов напрямую из базы данных TMDB/IMDB.

## Подготовка окружения
Импортируем разработанные аналитические классы из нашего собственного модуля.


```python
%timeit
from movielens_analysis import Movies, Ratings, Tags, Links
```

## Часть 1. Анализ каталога фильмов (Класс Movies)
Изучим общие характеристики фильмов: распределение по годам выпуска, популярность жанров и найдем самые "мультижанровые" картины.

```python
%%timeit -n 1 -r 1
movies = Movies("../latoyiah/movies.csv")
print("Распределение по годам выпуска:", movies.dist_by_release())
print("Популярность жанров в каталоге:", movies.dist_by_genres())
print("Топ-5 фильмов с наибольшим числом жанров:", movies.most_genres(5))
```

## Часть 2. Анализ оценок и активности пользователей (Класс Ratings и Users)
Перейдем к анализу поведения аудитории. Посмотрим, как распределялись оценки по годам и баллам, определим самые комментируемые фильмы, а также найдем наиболее спорные картины и категорию пользователей с полярными мнениями.

```python
%%timeit -n 1 -r 1
r = Ratings("../latoyiah/ratings.csv", "../latoyiah/movies.csv")
print("Распределение выставленных оценок по годам:", r.movies.dist_by_year())
print("Распределение оценок по шкале баллов:", r.movies.dist_by_rating())
print("Топ-5 фильмов по числу оценок:", r.movies.top_by_num_of_ratings(5))
print("Топ-5 лучших фильмов (Average):", r.movies.top_by_ratings(5, metric='average'))
print("Топ-5 лучших фильмов (Median):", r.movies.top_by_ratings(5, metric='median'))
print("Топ-5 самых спорных фильмов (Высокая дисперсия):", r.movies.top_controversial(5))
print("Распределение пользователей по числу оценок:", r.users.dist_by_num_of_ratings())
print("Распределение пользователей по средней оценке:", r.users.dist_by_avg_rating(metric='average'))
print("Топ-5 пользователей с самой нестабильной оценкой:", r.users.top_controversial_users(5))
```

## Часть 3. Лингвистический анализ тегов (Класс Tags)
Изучим текстовые маркеры (теги), которые пользователи оставляют к фильмам. Найдем самые популярные темы, самые длинные формулировки и теги, состоящие из наибольшего количества слов.

```python
%%timeit -n 1 -r 1
t = Tags("../delciefe/tags.csv")
print("Топ-5 тегов с наибольшим количеством слов:", t.most_words(5))
print("Топ-5 самых длинных тегов (по символам):", t.longest(5))
print("Пересечение длинных и многословных тегов:", t.most_words_and_longest(5))
print("Топ-5 самых популярных тегов:", t.most_popular(5))
print("Поиск уникальных тегов со словом 'sci-fi':", t.tags_with("sci-fi"))
```

## Часть 4. Интеграция с внешними данными IMDB (Класс Links)
В заключительной части мы задействуем веб-запросы к API, чтобы обогатить наш датасет финансовой и производственной информацией. Мы определим самых продуктивных режиссеров, самые дорогие, прибыльные и продолжительные фильмы.


```python
%%timeit -n 1 -r 1
l = Links("../delciefe/links.csv")
print("Информация из IMDB для фильмов [1, 2, 3]:", l.get_imdb([1, 2, 3], ["Title", "Director", "Budget", "Runtime"]))
print("Топ-3 режиссера по количеству фильмов:", l.top_directors(3))
print("Топ-3 самых высокобюджетных фильма:", l.most_expensive(3))
print("Топ-3 самых прибыльных фильма:", l.most_profitable(3))
print("Топ-3 самых продолжительных фильма:", l.longest(3))
print("Топ-3 фильма по стоимости производства одной минуты:", l.top_cost_per_minute(3))
```

## Выводы отчёта
1. На основе анализа каталога выделены ключевые исторические периоды активности кинематографа.
2. Изучены поведенческие паттерны пользователей: распределение оценок смещено в сторону положительных.
3. Лингвистический анализ тегов помог определить ключевые смысловые маркеры контента.
4. Внешняя интеграция предоставила точные экономические метрики (бюджет/окупаемость) для ключевых кинолент.
